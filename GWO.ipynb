```python
import numpy as np
import matplotlib.pyplot as plt

vehicles = np.array([25, 40, 15, 30])

cycle_time = 120
lower = 10
upper = 60

wolves = 20
iterations = 10

def fitness(green):
    congestion = np.sum((vehicles / green) ** 2)

    penalty = 5 * (np.sum(green) - cycle_time) ** 2

    return congestion + penalty


X = np.random.uniform(
    lower, upper, (wolves, 4)
)

alpha_score = float("inf")
beta_score = float("inf")
delta_score = float("inf")

alpha = np.zeros(4)
beta = np.zeros(4)
delta = np.zeros(4)

curve = []

for t in range(iterations):

    for i in range(wolves):

        X[i] = np.clip(X[i], lower, upper)

        score = fitness(X[i])

        if score < alpha_score:
            delta_score = beta_score
            delta = beta.copy()

            beta_score = alpha_score
            beta = alpha.copy()

            alpha_score = score
            alpha = X[i].copy()

        elif score < beta_score:
            delta_score = beta_score
            delta = beta.copy()

            beta_score = score
            beta = X[i].copy()

        elif score < delta_score:
            delta_score = score
            delta = X[i].copy()

    a = 2 - 2 * t / iterations

    for i in range(wolves):

        r1 = np.random.rand(4)
        r2 = np.random.rand(4)

        A1 = 2 * a * r1 - a
        C1 = 2 * r2
        D1 = abs(C1 * alpha - X[i])
        X1 = alpha - A1 * D1

        r1 = np.random.rand(4)
        r2 = np.random.rand(4)

        A2 = 2 * a * r1 - a
        C2 = 2 * r2
        D2 = abs(C2 * beta - X[i])
        X2 = beta - A2 * D2

        r1 = np.random.rand(4)
        r2 = np.random.rand(4)

        A3 = 2 * a * r1 - a
        C3 = 2 * r2
        D3 = abs(C3 * delta - X[i])
        X3 = delta - A3 * D3

        X[i] = np.clip(
            (X1 + X2 + X3) / 3,
            lower, upper
        )

    curve.append(alpha_score)

print("\nOPTIMIZED TRAFFIC SIGNAL TIMINGS")

for i in range(4):
    print(
        f"Road {i+1}: "
        f"{alpha[i]:.2f} seconds"
    )

print("\nTotal Green Time:", round(np.sum(alpha), 2))
print("Minimum Fitness:", round(alpha_score, 4))

normal = np.array([30, 30, 30, 30])

print("\nEqual Timing Fitness:", round(fitness(normal), 4))
print("GWO Optimized Fitness:", round(alpha_score, 4))

plt.figure(figsize=(8, 5))

plt.plot(curve, color="blue", linewidth=2)

plt.xlabel("Iterations")
plt.ylabel("Fitness Value")
plt.title("GWO Traffic Signal Optimization")

plt.grid()
plt.show()

plt.figure(figsize=(8, 5))

plt.bar(
    ["Road 1", "Road 2", "Road 3", "Road 4"],
    alpha,
    color=["red", "green", "orange", "blue"]
)

plt.xlabel("Traffic Roads")
plt.ylabel("Green Signal Time (seconds)")
plt.title("Optimized Traffic Signal Timings")

plt.show()


```
