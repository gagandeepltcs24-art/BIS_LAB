```python
import numpy as np

distances = np.array([
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
], dtype=float)

n = len(distances)
n_ants = 10
n_iterations = 10
alpha = 1
beta = 2
rho = 0.5
q = 100

pheromone = np.ones((n, n))
heuristic = 1 / (distances + 1e-10)
np.fill_diagonal(heuristic, 0)

best_tour = None
best_length = float('inf')

for iteration in range(n_iterations):
    tours = []
    lengths = []

    for ant in range(n_ants):
        start = np.random.randint(n)
        tour = [start]
        visited = {start}

        while len(tour) < n:
            current = tour[-1]

            probabilities = (
                pheromone[current] ** alpha
                * heuristic[current] ** beta
            )

            for city in visited:
                probabilities[city] = 0

            probabilities /= probabilities.sum()
            next_city = np.random.choice(n, p=probabilities)

            tour.append(next_city)
            visited.add(next_city)

        tour.append(start)

        length = sum(
            distances[tour[i]][tour[i + 1]]
            for i in range(n)
        )

        tours.append(tour)
        lengths.append(length)

        if length < best_length:
            best_length = length
            best_tour = tour.copy()

    pheromone *= (1 - rho)

    for tour, length in zip(tours, lengths):
        deposit = q / length

        for i in range(n):
            a, b = tour[i], tour[i + 1]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit
    print("Best Tour in ",iteration," iteration:",best_tour)
    print("Shortest Distance in ",iteration," iteration:",best_length)
print("Best Tour:", best_tour)
print("Shortest Distance:", best_length)
```
