```python
import random



jobs = [4, 7, 2, 5, 8, 3, 6, 4]

num_machines = 3

grid_size = 10
num_cells = grid_size * grid_size
iterations = 10

random.seed(42)


def calculate_makespan(schedule):
    machine_loads = [0] * num_machines

    for job_index, machine in enumerate(schedule):
        machine_loads[machine] += jobs[job_index]

    return max(machine_loads)



def fitness(schedule):
    return calculate_makespan(schedule)


def create_schedule():
    return [
        random.randint(0, num_machines - 1)
        for _ in jobs
    ]


def get_best_neighbor(grid, row, col):
    best_schedule = grid[row][col]

    best_fitness = fitness(best_schedule)

    for r in range(max(0, row - 1),
                   min(grid_size, row + 2)):

        for c in range(max(0, col - 1),
                       min(grid_size, col + 2)):

            neighbor = grid[r][c]
            neighbor_fitness = fitness(neighbor)

            if neighbor_fitness < best_fitness:
                best_schedule = neighbor
                best_fitness = neighbor_fitness

    return best_schedule


def mutate(schedule):
    new_schedule = schedule.copy()

    job_index = random.randint(0, len(jobs) - 1)

    new_schedule[job_index] = random.randint(
        0, num_machines - 1
    )

    return new_schedule


grid = [
    [
        create_schedule()
        for _ in range(grid_size)
    ]
    for _ in range(grid_size)
]


best_schedule = None
best_fitness = float("inf")



for iteration in range(iterations):

    new_grid = [
        [None for _ in range(grid_size)]
        for _ in range(grid_size)
    ]

    for row in range(grid_size):
        for col in range(grid_size):

            current_schedule = grid[row][col]

            current_fitness = fitness(current_schedule)

            neighbor_schedule = get_best_neighbor(
                grid, row, col
            )

            candidate = mutate(neighbor_schedule)

            candidate_fitness = fitness(candidate)

            if candidate_fitness < current_fitness:
                new_grid[row][col] = candidate
            else:
                new_grid[row][col] = current_schedule

            cell_fitness = fitness(new_grid[row][col])

            if cell_fitness < best_fitness:
                best_fitness = cell_fitness
                best_schedule = new_grid[row][col].copy()

    grid = new_grid

    print(
        f"Iteration {iteration + 1}: "
        f"Best Makespan = {best_fitness}"
    )




print("\nBest Job Schedule")
print("-----------------")

machine_loads = [0] * num_machines

for job_index, machine in enumerate(best_schedule):
    machine_loads[machine] += jobs[job_index]

for machine in range(num_machines):
    assigned_jobs = [
        f"J{job_index + 1}"
        for job_index, assigned_machine
        in enumerate(best_schedule)
        if assigned_machine == machine
    ]

    print(
        f"Machine {machine + 1}: "
        f"{assigned_jobs}, "
        f"Total Processing Time = "
        f"{machine_loads[machine]}"
    )

print("\nMinimum Makespan:", best_fitness)
```
